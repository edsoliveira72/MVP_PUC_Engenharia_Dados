### Pergunta 2: Top 5 estados com maiores índices de atraso na entrega (%)

```python
# Pergunta 1: Tempo médio de entrega por estado do cliente
# Pergunta 2: Top 5 estados com maiores índices de atraso na entrega (%)
query_logistica = """
SELECT 
    customer_state AS `Estado de Destino`,
    tempo_medio_entrega_dias AS `Tempo Médio de Entrega (dias)`,
    pct_pedidos_atrasados AS `Índice de Atraso (%)`,
    total_pedidos_entregues,
    total_pedidos_atrasados
FROM gold_olist.gold_metrics_logistica
ORDER BY `Índice de Atraso (%)` DESC
"""

df_logistica = spark.sql(query_logistica)
display(df_logistica)
```

### Análise do resultado:

#### Resposta 1 (Tempo Médio de Entrega): Existe uma forte assimetria geográfica. Enquanto no estado de São Paulo a entrega ocorre em média em 8,66 dias, nos estados das regiões Norte e Nordeste o prazo ultrapassa 20 a 24 dias corridos.

#### Resposta 2 (Top 5 Estados em Atraso): Alagoas (24,12%), Maranhão (20,38%), Sergipe (16,27%), Piauí (15,49%) e Ceará (15,29%) lideram os descumprimentos de prazo. Nessas localidades, 1 em cada 5 pedidos é entregue fora do prazo estimado.
